# 04 · ChatGPT Few-Shot Classification

Classify companies along UVP and Data Uniqueness with the OpenAI API in a few-shot setting: the model is primed with eight labelled example companies, then classifies the rest (thesis §3.6, Appendix B).

> Requires `OPENAI_API_KEY` in a local `.env` file (see `.env.example`).

```python
import pandas as pd

from company_assessment import config
from company_assessment.models.chatgpt import get_client, classify_dataframe
```

### Load the few-shot examples and the companies to classify

```python
examples = pd.read_csv(config.EXAMPLES_CSV)
examples = examples.loc[:, ~examples.columns.str.contains('^Unnamed')]
examples['Data Uniqueness'] = examples['Data Uniqueness'].fillna('N/A')

final_df = pd.read_csv(config.FINAL_CHATGPT_CSV)[:104]
len(examples), len(final_df)
```

### Classify

Swap `model_name` for `gpt-3.5-turbo-0125`, `gpt-4o`, or `gpt-4-turbo`.

```python
client = get_client()
model_name = 'gpt-4-turbo'

classified = classify_dataframe(client, final_df, examples, model_name=model_name)
classified[['UVP_predicted', 'Data Uniqueness_predicted']].head()
```

### Save predictions

```python
out_path = config.PROCESSED_DIR / f'final_classified_{model_name}.csv'
classified.to_csv(out_path, index=False)
print('saved', out_path)
```
