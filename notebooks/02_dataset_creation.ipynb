# 02 · Dataset Creation

Build the evaluation dataset from the Y Combinator Directory dump (thesis §3.1):

1. Load the raw YC companies CSV.
2. Keep companies tagged *Artificial Intelligence*.
3. Scrape each landing page into a `Scraped_Info` column.
4. Filter out companies with too little scraped text.

> Scraping hits live websites, so this step is slow and non-deterministic. Outputs land in `data/interim/` (git-ignored).

```python
import pandas as pd

from company_assessment import config
from company_assessment.data import build_scraped_ai_dataset, filter_by_scraped_length
```

### Load the raw YC directory

```python
ycd = pd.read_csv(config.YC_COMPANIES_CSV)
ycd.shape
```

### Filter to AI companies and scrape their landing pages

```python
scraped_ai = build_scraped_ai_dataset(ycd)
config.SCRAPED_AI_CSV.parent.mkdir(parents=True, exist_ok=True)
scraped_ai.to_csv(config.SCRAPED_AI_CSV, index=False)
scraped_ai.shape
```

### Keep companies with enough scraped text

```python
filtered_data = filter_by_scraped_length(scraped_ai)
filtered_data.to_csv(config.FILTERED_DATA_CSV, index=False)
filtered_data.shape
```
